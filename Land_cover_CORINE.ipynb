Code for calculating areas of different land types for N fixation calculations

Run from gis_env

```python
import rasterio
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load the CORINE Land Cover raster data
raster_path = 'path/to/CORINE_Land_Cover_2018_Norway.tif'  # Update this with the actual path
with rasterio.open(raster_path) as src:
    land_cover = src.read(1)  # Read the first band
    profile = src.profile  # Get the profile for metadata

# Check the unique classes in the raster
unique_classes = np.unique(land_cover)
print("Unique Land Cover Classes:", unique_classes)

# Define the pixel area (100m x 100m = 10,000 m² = 1 hectare)
pixel_area_hectares = (100 * 100) / 10000  # Convert to hectares

# Create an empty DataFrame to store the area per land cover class
area_summary = pd.DataFrame(columns=['Land Cover Class', 'Area (hectares)'])

# Calculate area for each land cover class
for lc_class in unique_classes:
    if lc_class > 0:  # Assuming that 0 represents no data
        pixel_count = np.count_nonzero(land_cover == lc_class)  # Count pixels for class
        area_hectares = pixel_count * pixel_area_hectares  # Calculate area
        area_summary = area_summary.append({'Land Cover Class': lc_class, 'Area (hectares)': area_hectares}, ignore_index=True)

# Optional: Sort by area
area_summary.sort_values(by='Area (hectares)', ascending=False, inplace=True)

# Output the area summary
print("Area Summary by Land Cover Class:")
print(area_summary)

# Optional: Visualize the area by land cover
plt.figure(figsize=(10, 6))
plt.barh(area_summary['Land Cover Class'].astype(str), area_summary['Area (hectares)'], color='skyblue')
plt.xlabel('Area (hectares)')
plt.title('Area by Land Cover Class in Norway (CORINE Land Cover 2018)')
plt.show()
```
